### سوال ۱: مسئله اصلی کسب‌وکار را با زبان ساده توضیح دهید.

فروشگاه آنلاین بین‌المللی متوجه شده بخشی از مشتریان قدیمی به‌تدریج خریدشان را کم می‌کنند و بخشی بی‌صدا قطع می‌کنند. شرکت فقط **بعد از رفتن** مشتری متوجه می‌شود. می‌خواهیم از روی الگوی خرید ۱۸۰ روز گذشته (فاصله از آخرین خرید، افت تعداد سفارش، کاهش مبلغ، نرخ لغو و مرجوعی، تنوع سبد) مشتریان در معرض ریزش را **قبل از رفتن** پیدا کنیم و با بودجه محدود **۵۰۰ تماس**، مناسب‌ترین اقدام حفظ مشتری را برای هر گروه طراحی کنیم.

### سوال ۷: چه شاخص‌هایی نشان می‌دهد پروژه موفق بوده است؟ (KPI)

**شاخص‌های خوشه‌بندی:** Silhouette قابل قبول، تفسیرپذیری تجاری خوشه‌ها، تفاوت معنادار نرخ ریزش بین خوشه‌ها، پایداری نسبی در تاریخ‌های مرجع.
**شاخص‌های طبقه‌بندی:** PR-AUC، Recall، F1، Precision@500، بهتر از خط مبنای RFM، ارزیابی زمانی بدون نشت.
**شاخص‌های کسب‌وکاری:** کاهش نرخ ریزش در گروه کمپین، سود خالص مثبت، نرخ موفقیت بالاتر از سربه‌سر، عملکرد بهتر از انتخاب تصادفی.

### سوال ۸: حداقل پنج KPI.

1. **PR-AUC** مدل برتر روی آزمون (اوت+سپتامبر).
2. **Precision@500** (دقت در ۵۰۰ نفر اول لیست ریسک).
3. **Recall** در آستانه تصمیم انتخاب‌شده.
4. **نرخ سربه‌سر کمپین** = `C_contact / V_retained`.
5. **سود خالص مورد انتظار کمپین** = (مشتریان حفظ‌شده × ارزش) − (۵۰۰ × هزینه تماس).
6. (اختیاری) **Silhouette Score** خوشه‌بندی و **تفاوت نرخ ریزش بین خوشه‌ها**.

### سوال ۹: موفقیت این پروژه از دید مدیر فروشگاه چگونه اندازه‌گیری می‌شود؟

- **کاهش نرخ ریزش** گروه کمپین در مقایسه با گروه کنترل (نیازمند A/B Test).
- **سود خالص مثبت** کمپین با فرض‌های شفاف.
- **نرخ موفقیت بالاتر از سربه‌سر** در سناریوهای مختلف هزینه/ارزش.
- **بازده بهتر ۵۰۰ تماس نسبت به انتخاب تصادفی** (Precision@500 بالاتر از نرخ پایه).
- **گزارش قابل دفاع** با ذکر محدودیت‌ها و نبود اثبات علّی.

### سوال ۱۰: موفقیت این پروژه از دید تیم بازاریابی/CRM چگونه اندازه‌گیری می‌شود؟

- **کیفیت لیست ۵۰۰ نفره:** مشتریان واقعاً پرریسک و قابل هدف‌گیری باشند.
- **تفکیک اقدام:** هر گروه پیام مناسب خودش را دریافت کند (`Action` متمایز).
- **نرخ پاسخ/بازگشت** در گروه‌های مختلف (پس از اجرای A/B Test).
- **کاهش هزینه تماس‌های بی‌نتیجه** نسبت به روش تصادفی.
- **قابلیت اجرا:** خروجی با CustomerID هش‌شده و ستون اقدام، بدون نیاز به بازسازی.

### سوال ۱۱: اگر سیستم ۹۰٪ دقت داشته باشد اما تیم بازاریابی به آن اعتماد نکند، آیا پروژه موفق است؟ چرا؟

**خیر.** سه دلیل:
1. **دقت ۹۰٪ به‌تنهایی معنادار نیست** وقتی نرخ پایه ریزش بالاست؛ مدلی که همه را «ریزشی» بگوید Accuracy بالایی دارد ولی تفکیک نمی‌کند. به همین دلیل در پروژه ما Accuracy حذف و PR-AUC و Precision@500 جایگزین شد.
2. **اعتماد تیم اجرایی شرط لازم است:** اگر تیم بازاریابی لیست را باور نکند، آن را اجرا نمی‌کند و اثر تجاری صفر می‌شود.
3. **تفسیرپذیری:** بدون توضیح عوامل مؤثر (Permutation Importance + PDP) و بدون شفافیت روش، اعتماد ساخته نمی‌شود.

### سوال ۱۲: مزایای مالی احتمالی پروژه.

- حفظ درآمد آینده مشتریان پرارزش که در آستانه ریزش‌اند.

### سوال ۱۳: مزایای عملیاتی احتمالی پروژه.

### سوال ۲۰: محدودیت‌های مالی.

- **بودجه ثابت ۵۰۰ تماس:** نمی‌توان به همه ریزشی‌های بالقوه رسید.
- **هزینه تماس متغیر:** با افزایش `C_contact`، نرخ سربه‌سر `C_contact / V_retained` سخت‌تر می‌شود.
- **هزینه تخفیف:** در سناریوهای پرهزینه، آستانه سربه‌سر به سطح غیرواقعی می‌رسد (مثلاً هزینه ۱۵ و ارزش ۵۰ → نرخ موفقیت لازم ۳۰٪).
- **هزینه فرصت:** هر تماس به مشتری اشتباه، یک تماس از دست رفته برای مشتری واقعی است.
- **نبود بودجه A/B Test در برخی سازمان‌ها** که اثبات علّی را غیرممکن می‌کند.

### سوال ۲۱: محدودیت‌های فنی.

- **نبود `openpyxl`:** خواندن `.xlsx` ممکن نیست؛ راه‌حل: `fetch_ucirepo(id=352)` یا تبدیل یک‌باره به CSV.
- **نبود `xgboost`:** جایگزین `HistGradientBoostingClassifier` (خانواده Gradient Boosting با `class_weight`).
- **نبود `shap`:** جایگزین Permutation Importance + `feature_importances_` + PDP.
- **نوع ستون `InvoiceDate` از `fetch_ucirepo`:** رشته است و باید `pd.to_datetime` اعمال شود؛ در غیر این صورت `TypeError: '<=' not supported between Timestamp and str`.
- **عدم توازن کلاس:** `class_weight='balanced'` و معیار PR-AUC.
- **نشت اطلاعات:** مقیاس‌گذاری داخل Pipeline، ویژگی‌ها فقط از ۱۸۰ روز قبل، برچسب فقط از ۹۰ روز بعد.
- **حجم داده متوسط (~۵۴۱ هزار رکورد):** K-Means با `sample_size=5000` برای Silhouette و `n_init=10`.
- **قابلیت بازتولید:** seed ثابت `RANDOM_STATE=42`.

## بخش سوم: Determine Data Mining Goals (سوالات ۳۱–۵۰)

### سوال ۳۱: هدف داده‌کاوی این پروژه چیست؟

ساخت یک سامانه دو مرحله‌ای:
1. **خوشه‌بندی K-Means** روی ۱۴ ویژگی رفتاری برای کشف گروه‌های طبیعی مشتریان و سنجش تفاوت نرخ ریزش آن‌ها.
2. **طبقه‌بندی دودویی** با استفاده از ویژگی‌های رفتاری + برچسب خوشه برای پیش‌بینی احتمال نداشتن خرید معتبر در ۹۰ روز آینده و رتبه‌بندی مشتریان برای انتخاب ۵۰۰ نفر.

### سوال ۳۲: تفاوت Business Goal و Data Mining Goal.

| | Business Goal | Data Mining Goal |
|---|---|---|
| زبان | مدیر/کسب‌وکار | دانشمند داده/فنی |
| سوال | «چه نتیجه‌ای می‌خواهیم؟» | «مدل دقیقاً چه چیزی را پیش‌بینی کند؟» |
| اندازه‌گیری | سود، نرخ ریزش، هزینه | PR-AUC، Recall، F1، Precision@500 |
| در این پروژه | کاهش ریزش و سود مثبت کمپین | پیش‌بینی ریزش ۹۰ روزه با طبقه‌بندی |
| خروجی | تصمیم و اقدام | مدل و پیش‌بینی |

### سوال ۳۳: هدف داده‌کاوی در یک جمله.

> **خوشه‌بندی رفتار مشتریان با K-Means، سپس پیش‌بینی احتمال ریزش هر مشتری (نداشتن خرید معتبر در ۹۰ روز پس از تاریخ مرجع) با استفاده از ویژگی‌های ۱۸۰ روز قبل، و رتبه‌بندی مشتریان بر اساس این احتمال.**

### سوال ۳۴: Classification, Regression, Clustering یا Association Rule؟

**پاسخ اصلی: Classification دودویی، اما مسیر از Clustering می‌گذرد.**

| نوع | مناسب؟ | دلیل |
|---|---|---|
| **Clustering** | ✅ بله، **گام اول** | K-Means روی ۱۴ ویژگی رفتاری، K بهینه ۴ تا ۶، پروفایل خوشه، نرخ ریزش هر خوشه |
| **Classification** | ✅ بله، **گام دوم** | خروجی دسته دودویی: ریزش=۱ / عدم ریزش=۰؛ برچسب مشتق‌شده با قاعده ۹۰ روزه |
| Regression | ❌ نه (در هسته) | هدف عدد پیوسته نیست (اگر «مبلغ خرید آینده» بود، رگرسیون بود) |
| Association Rule | ❌ نه | هدف کشف قواعد سبد خرید نیست |

**ترتیب اجرایی:** Clustering → Classification → (اختیاری) Clustering ثانویه روی پرریسک‌ها برای طراحی اقدام دقیق‌تر.

### سوال ۳۵: اگر خروجی پروژه احتمال ریزش مشتری باشد، نوع مسئله چیست؟

**Classification دودویی (با خروجی احتمال).** مدل یک عدد بین ۰ و ۱ به‌عنوان `RiskScore` تولید می‌کند که سپس با آستانه تصمیم (تنظیم‌شده روی داده آموزش) به کلاس ۰/۱ تبدیل می‌شود. در پروژه ما `best_threshold_on_train` این کار را با هدف بیشینه‌سازی F1 انجام می‌دهد.

### سوال ۳۶: اگر خروجی پروژه «امتیاز ریسک پیوسته» یا «ارزش آینده مشتری (CLV)» باشد، نوع مسئله چیست؟

**Regression.** اگر هدف پیش‌بینی یک عدد پیوسته (مثل CLV یا مبلغ خرید ۹۰ روز آینده) باشد، رگرسیون است. در پروژه فعلی این کار انجام نشد، اما به‌عنوان **گسترش آینده** پیشنهاد می‌شود: پیش‌بینی CLV به‌عنوان ویژگی کمکی برای اولویت‌بندی کمپین.

### سوال ۳۷: اگر خروجی پروژه گروه‌بندی مشتریان باشد، نوع مسئله چیست؟

**Clustering (بدون نظارت).** این همان **گام اول پروژه ما** است: K-Means روی ۱۴ ویژگی رفتاری با `StandardScaler` و `log1p` برای کاهش چولگی. K بهینه با Elbow و Silhouette انتخاب شد و خوشه‌ها با نام‌های تجاری (Champions، Loyalists، Need Attention، At-Risk، Hibernating) نگاشت شدند.

### سوال ۳۸: چه معیارهایی برای ارزیابی مدل اهمیت دارند؟

| معیار | کاربرد |
|---|---|
| Silhouette / Elbow | ارزیابی خوشه‌بندی |
| تفاوت نرخ ریزش بین خوشه‌ها | ارزیابی معناداری خوشه‌بندی |
| Precision | هزینه تماس اشتباه |
| Recall | مشتریان از دست رفته |
| F1 | تعادل Precision/Recall |
| **PR-AUC** | **معیار اصلی طبقه‌بندی** |
| ROC-AUC | مکمل PR-AUC |
| Confusion Matrix | نمایش دقیق خطا |
| **Precision@500** | **مستقیماً مرتبط با تصمیم کمپین** |

### سوال ۳۹: Accuracy چیست و آیا کافی است؟

Accuracy = پیش‌بینی‌های درست ÷ کل. **کافی نیست.** اگر نرخ پایه ریزش حدود ۶۵–۷۰٪ باشد (که در پنجره‌های ۹۰ روزه دور از انتظار نیست)، مدلی که «همه را ریزشی» بگوید Accuracy بالایی می‌گیرد ولی تفکیک انجام نمی‌دهد. در پروژه ما Accuracy از گزارش حذف و PR-AUC، Recall، F1 و Precision@500 جایگزین شد.

### سوال ۴۰: Precision چیست؟

از میان مشتریانی که مدل «ریزشی» پیش‌بینی کرده، چند درصد واقعاً ریزشی بودند. `Precision = TP / (TP + FP)`.

### سوال ۴۱: Recall چیست؟

از میان مشتریان واقعاً ریزشی، چند درصد را مدل پیدا کرد. `Recall = TP / (TP + FN)`.

### سوال ۴۲: Precision مهم‌تر است یا Recall؟ چرا؟

**بستگی به سطح تصمیم دارد:**
- **ارزیابی کلی مدل:** Recall و PR-AUC مهم‌ترند، چون از دست دادن مشتری ارزشمند گران‌تر از یک تماس اضافی است.
- **انتخاب ۵۰۰ نفر با بودجه ثابت:** **Precision@500** معیار عملیاتی اصلی است، چون هر تماس به مشتری اشتباه یعنی یک فرصت از دست رفته.
- **انتخاب نهایی آستانه:** در تحلیل حساسیت با فرمول `p_success > C_contact / V_retained` تصمیم گرفته می‌شود.

### سوال ۴۳: هزینه False Positive چیست؟

مشتری‌ای که نمی‌رفت، پرریسک اعلام شد و تماس/تخفیف گرفت:
- هزینه تماس (`C_contact`).
- هزینه تخفیف احتمالی.
- اشغال ظرفیت ۵۰۰ تماس که می‌توانست به مشتری واقعی برسد.
- احتمال مزاحمت برای مشتری وفادار.

### سوال ۴۴: هزینه False Negative چیست؟

مشتری‌ای که می‌رفت، پیدا نشد:
- از دست دادن درآمد آینده مشتری (`V_retained`).
- از دست دادن فرصت حفظ با یک اقدام کم‌هزینه.
- کاهش اعتماد مدیریت به مدل.

### سوال ۴۵: کدام خطا برای فروشگاه خطرناک‌تر است؟

**False Negative** معمولاً خطرناک‌تر است، چون از دست دادن یک مشتری ارزشمند گران‌تر از یک تماس اضافی است. **اما** با قید بودجه ثابت ۵۰۰ تماس، هر FP یک فرصت از دست رفته است؛ بنابراین در **انتخاب ۵۰۰ نفر**، Precision@500 اهمیت ویژه دارد. تعادل نهایی به `C_contact` و `V_retained` بستگی دارد و در ماتریس حساسیت بررسی شد.

### سوال ۴۶: موفقیت مدل را چگونه تعریف می‌کنید؟

مدل موفق است اگر **همه** شرایط برقرار باشد:
1. از خط مبنای RFM (`Recency >= 90`) در PR-AUC و Precision@500 بهتر باشد.
2. از انتخاب تصادفی ۵۰۰ نفر بهتر باشد.
3. روی داده آینده (اوت+سپتامبر) ارزیابی شده باشد.
4. بدون نشت اطلاعات.
5. در تاریخ‌های مرجع مختلف پایدار باشد.
6. عوامل مؤثر قابل تفسیر باشند (Permutation + PDP).
7. با فرض‌های شفاف، سود مورد انتظار کمپین مثبت باشد.

### سوال ۴۷: حداقل سه معیار فنی موفقیت مدل.

1. **PR-AUC مدل برتر > PR-AUC خط مبنای RFM** روی آزمون اوت+سپتامبر.
2. **Precision@500 بالاتر از نرخ پایه** و بهتر از انتخاب تصادفی.
3. **پایداری در تاریخ‌های مرجع مختلف:** PR-AUC در ژوئن، ژوئیه، اوت، سپتامبر تفاوت چشمگیری نداشته باشد.
4. **Ablation:** نشان دهد برچسب خوشه یا ویژگی‌های تغییر رفتار ارزش افزوده واقعی دارند.
5. **بدون نشت:** هیچ ویژگی‌ای از پنجره برچسب استفاده نکند.

### سوال ۴۸: حداقل سه معیار کسب‌وکاری موفقیت پروژه.

1. **کاهش نرخ ریزش** گروه کمپین نسبت به گروه کنترل (A/B Test).
2. **سود خالص مثبت** با فرض‌های شفاف.
3. **نرخ موفقیت کمپین بالاتر از نرخ سربه‌سر** در سناریوهای مختلف هزینه/ارزش.
4. **رضایت تیم بازاریابی** از کیفیت لیست و تفکیک اقدام.
5. **قابلیت اجرا و بازتولید** خروجی در دوره‌های بعدی.

### سوال ۴۹: Data Mining Goal Statement رسمی.
